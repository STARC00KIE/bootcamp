```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# 캔버스 사이즈 조정
fig, ax = plt.subplots(figsize=(10.73, 11.14))
ax.set_xlim(0, 10)
ax.set_ylim(0, 12)
ax.axis('off')

# 파이프라인 단계 정의
stages = [
    "중고차 이미지 수집 및 탐색",
    "Albumentations 기반 데이터 전처리",
    "CNN / Transformer 모델 설계 및 학습\n(LogLoss 기반 성능 비교)",
    "Tableau를 통한 결과 시각화 및 분석\n(AvgLoss = 0.3)",
    "코드 리팩토링 및 GitHub 문서화"
]

# 박스 위치 설정
box_width = 8
box_height = 1.2
y_positions = [10, 8, 6, 4, 2]

# 박스와 텍스트 그리기
for i, (stage, y) in enumerate(zip(stages, y_positions)):
    rect = patches.FancyBboxPatch((1, y), box_width, box_height,
                                   boxstyle="round,pad=0.05",
                                   edgecolor="black", facecolor="#DCE6F1")
    ax.add_patch(rect)
    ax.text(1 + box_width / 2, y + box_height / 2, stage,
            ha='center', va='center', fontsize=20)

    # 화살표
    if i < len(stages) - 1:
        ax.annotate('', xy=(5, y-0.1), xytext=(5, y - 0.7),
                    arrowprops=dict(arrowstyle='<-', lw=1.5))

plt.tight_layout()
# plt.savefig("/mnt/data/used_car_pipeline_diagram.png", dpi=300)
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import textwrap

# 전처리 파이프라인 코드 (문자열)
code = """
def get_train_transform(img_size):
    \"\"\"
    학습 데이터용 데이터 증강 파이프라인
    - 조명, 방향, 위치, 그림자 등 다양한 상황에 대응
    \"\"\"
    return A.Compose([
        A.Resize(img_size, img_size),

        # 좌우 반전: 차량 방향 다양화
        A.HorizontalFlip(p=0.5),

        # 밝기/대비 랜덤 조정
        A.RandomBrightnessContrast(p=0.2),

        # 색상(Hue), 채도(Saturation), 명도(Value)
        A.HueSaturationValue(p=0.2),

        # 이동, 확대/축소, 회전
        A.ShiftScaleRotate(shift_limit=0.05,
                           scale_limit=0.05,
                           rotate_limit=15,
                           p=0.5),

        # 그림자 추가
        A.RandomShadow(p=0.2),

        # RGB 색상 채널 변형
        A.RGBShift(r_shift_limit=10, g_shift_limit=10, b_shift_limit=10, p=0.2),

        # 일부분 무작위 제거 (가림 상황)
        A.CoarseDropout(max_holes=2,
                        max_height=16,
                        max_width=16,
                        p=0.3),

        # 정규화 (ImageNet mean/std)
        A.Normalize(mean=(0.485, 0.456, 0.406),
                    std=(0.229, 0.224, 0.225)),

        ToTensorV2()
    ])
"""

# 시각화
fig, ax = plt.subplots(figsize=(15.07 / 2.54, 8.94 / 2.54))  # cm to inches
wrapped_code = "\n".join(textwrap.wrap(code, width=100))
ax.text(0, 1, wrapped_code, fontsize=10, fontfamily='monospace', va='top')
ax.axis('off')

# 저장
#output_path = "/mnt/data/train_transform_pipeline_resized.png"
# plt.savefig(output_path, bbox_inches='tight', dpi=300)
plt.show()
```
