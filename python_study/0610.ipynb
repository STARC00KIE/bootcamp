```python
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
print(cancer.DESCR)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(cancer.data,
                                                    cancer.target,
                                                    stratify= cancer.target,
                                                    random_state = 7)

x_train.shape, y_train.shape
```

```python
# 3. 모델 생성 및 학습

from sklearn.naive_bayes import GaussianNB

cancer_nb = GaussianNB()

cancer_nb.fit(x_train, y_train)

pred = cancer_nb.predict(x_test)
```

```python
# 4. 예측 확인
print(y_test)    # 정답
print(pred)      # 예측값
print(np.mean(y_test == pred))   # 예측 정답률
```

```python
# 데이터 확인
data = cancer.data
target = cancer.target
print(data)
print(target)
```

```python
# 7:3 비율로 데이터 스플릿
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(data,
                                                    target,
                                                    test_size=0.3,
                                                    random_state=0)

print(x_train.shape, y_test.shape)
```

```python
# 고유값 출력
np.unique(y_test, return_counts=True)
```

```python
print(np.min(x_train, axis=0))
print(np.max(x_train, axis=0))
print(np.mean(x_train, axis=0))
print(np.std(x_train, axis=0))
```

```python
(x_train - np.mean(x_train, axis=0)) / np.std(x_train, axis=0)
```

```python
# STANDARD SCALER 적용
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train)
x_train_zst = scaler.transform(x_train)
x_test_zst = scaler.transform(x_test)
print(x_train[:5])
print(x_train_zst[:5])
```

```python
# KNN 알고리즘 적용
from sklearn.neighbors import KNeighborsClassifier
cancer_knn = KNeighborsClassifier(n_neighbors=10)
cancer_knn.fit(x_train_zst, y_train)
pred = cancer_knn.predict(x_test_zst)
```

```python
# 실제 라벨값과 예측값 출력
print(y_test)
print(pred)
```

```python
#평균 정확도 출력
print(np.mean(y_test == pred))
```

```python
# K 개수 변하면서 예측값 확률 변화 확인인
for k in range(1,21):
    cancer_knn = KNeighborsClassifier(n_neighbors=k)
    cancer_knn.fit(x_train_zst, y_train)
    pred = cancer_knn.predict(x_test_zst)
    print(k, "==>", np.mean(y_test == pred))
```

```python
# 마찬가지로 진행
cancer_knn = KNeighborsClassifier(n_neighbors=13)
cancer_knn.fit(x_train_zst, y_train)
pred = cancer_knn.predict(x_test_zst)
print(k, "==>", np.mean(y_test == pred))
```

```python
# 0. 필요 라이브러리 임포트
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import StratifiedShuffleSplit

# 1. 필요 변수 세팅

# 1.1 공통
split = StratifiedShuffleSplit(n_splits=10, test_size=0.3, random_state=7)         # 데이터 반복 분할
first = True                                     # 첫 루프에서만 나이브베이즈 모델에 대한 구문에 들어가기 위한.. 꼼수? 방법.

# 1.2. 나이브 베이즈 모델에 쓸 것
nb_cv_acc = []                                   # nb 정확도 저장용 리스트

# 1.3. kNN 모델에 쓸 것
scaler = StandardScaler()                        # 표준화 도구 생성
cancer_zst = scaler.fit_transform(cancer.data)   # z-score 표준화 한 data를 새 변수에 저장
knn_cv_acc = []                                  # knn 정확도 저장용 리스트

# ------------------------------------------------

# 2. 분할된 데이터로 반복 모델링하여 평균 정확도를 저장 + 비교하기 위한 반복문 작성

for k in range(1, 41, 2):      # 적절한 k값을 찾을 범위 지정
  tmp_acc = []                 # 현재 k값에 대한 정확도를 저장할 리스트 for 최적의 k값 찾기

  for train_idx, test_idx in split.split(cancer.data, cancer.target):

    if first:
      cancer_nb = GaussianNB()
      cancer_nb.fit(cancer.data[train_idx], cancer.target[train_idx])
      pred = cancer_nb.predict(cancer.data[test_idx])
      nb_cv_acc.append(np.mean(cancer.target[test_idx] == pred))

    cancer_knn = KNeighborsClassifier(n_neighbors=k)
    cancer_knn.fit(cancer_zst[train_idx], cancer.target[train_idx])
    pred = cancer_knn.predict(cancer_zst[test_idx])
    tmp_acc.append(np.mean(cancer.target[test_idx] == pred))

  knn_cv_acc.append(np.mean(tmp_acc))
  first = False

print(np.mean(nb_cv_acc))
print(np.argmax(knn_cv_acc), knn_cv_acc[np.argmax(knn_cv_acc)])
```

# GridSearchCV 사용
- 하이퍼파라미터를 자동으로 튜닝해주는 기능

```python
from sklearn.model_selection import GridSearchCV

# 0. 파라미터 범위 및 목록 지정
param_grid = [{'n_neighbors':range(1,41,2)}]

# 1. 필요한 변수
scaler = StandardScaler()   # 표준화
cancer_zst = scaler.fit_transform(cancer.data)

# 2. 모델 생성
cancer_knn = KNeighborsClassifier()

# 3. 그리드서치
grid_m = GridSearchCV(cancer_knn,     # 사용할 모델 지정
                      param_grid,     # 사용할 파라미터 지정
                      cv=10,          # 시행할 교차검증 횟수 지정
                      return_train_score=True,    # 학습 정확도 저장 여부 지정
                      verbose=3)      # 진행상황 출력

grid_m.fit(cancer_zst, cancer.target)
```

```python
from sklearn.pipeline import Pipeline
pip = Pipeline([("preprocess", StandardScaler()),
                ("classifier", KNeighborsClassifier())])
grid_param = [
    {"preprocess":[StandardScaler()],
     "classifier":[KNeighborsClassifier()],
     "classifier__n_neighbors":range(1,40,2)},
    {"preprocess":[None],
     "classifier":[GaussianNB()]}
]

grid_m = GridSearchCV(pip, grid_param, cv=5,
                      return_train_score=True, verbose=3)
grid_m.fit(cancer.data, cancer.target)
```
