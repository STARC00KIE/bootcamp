```python
from sklearn.datasets import load_diabetes
dibt = load_diabetes()

print(dibt.DESCR)

# dibt 데이터는 수치적인 속성들을 여럿 가지고 있음
```

```python
dibt.data.shape
```

```python
# 예측하고자 하는 값 확인

dibt.target
```

```python
# 데이터 분할

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(dibt.data, dibt.target, random_state = 3)

print(x_train.shape, y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
dibt_lr = LinearRegression()
dibt_lr.fit(x_train, y_train)

pred = dibt_lr.predict(x_test)

print(pred[:10])
print(y_test[:10])
```

```python
print(dibt_lr.coef_)        # 회귀 계수
print(dibt_lr.intercept_)   # y절편
```

```python
x_test[0]    # 샘플 하나 확인
```

```python
# 모델의 예측값 계산 (y = wx + b)

import numpy as np
np.sum(x_test[0] * dibt_lr.coef_) + dibt_lr.intercept_
```

```python
# 성능평가

from sklearn.metrics import mean_squared_error, r2_score
print(mean_squared_error(y_test, dibt_lr.predict(x_test)))
print(r2_score(y_test, dibt_lr.predict(x_test)))
```

```python
## kNN 회귀, 결정트리회귀

from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor

dibt_knn = KNeighborsRegressor(n_neighbors=5)
dibt_knn.fit(x_train, y_train)

pred = dibt_knn.predict(x_test)

print(pred[:10])
print(y_test[:10])

# 평가
from sklearn.metrics import mean_squared_error, r2_score
print(mean_squared_error(y_test, pred))    # MSE
print(r2_score(y_test, pred))              # R2
```

```python
# 하이퍼파라미터 조정 : k값
# 여러 k값에 따른 MSE, R2 점수를 출력하여, 최적의 k값을 찾아보자

# 참고) MSE는 작은수록 좋고, R2는 1에 가까울수록 좋음

for k in range(1, 100, 2):
  dibt_knn = KNeighborsRegressor(n_neighbors=k)
  dibt_knn.fit(x_train, y_train)
  pred = dibt_knn.predict(x_test)
  print(k, mean_squared_error(y_test, pred), r2_score(y_test, pred))
```

```python
dibt_dt = DecisionTreeRegressor()
dibt_dt.fit(x_train, y_train)
pred = dibt_dt.predict(x_test)
print(mean_squared_error(y_test, pred))
```

```python
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor
dibt_rf = RandomForestRegressor()
dibt_rf.fit(x_train, y_train)
pred = dibt_rf.predict(x_test)
print(mean_squared_error(y_test, pred))
```

```python
dibt_ab = AdaBoostRegressor()
dibt_ab.fit(x_train, y_train)
pred = dibt_ab.predict(x_test)
print(mean_squared_error(y_test, pred))
```
