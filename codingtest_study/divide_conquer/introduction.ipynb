# 이진 탐색
- 정렬된 배열에서 특정 값을 중간 값을 기준으로 나누어 탐색
- 탐색 범위 반으로 줄임 -> 시간복잡도 O(log n)
- 저번에 했던것

```python
def binary_search(arr, target):
    start, end = 0, len(arr) - 1  # 시작 인덱스와 끝 인덱스 초기화

    while start <= end:  # 탐색 범위가 유효한 동안 반복
        mid = (start + end) // 2  # 중간 인덱스 계산
        if arr[mid] == target:    # 찾는 값이면 해당 인덱스 반환
            return mid
        elif arr[mid] < target:   # 타겟이 오른쪽에 있다면 시작점을 이동
            start = mid + 1
        else:                     # 타겟이 왼쪽에 있다면 끝점을 이동
            end = mid - 1

    return -1  # 못 찾은 경우 -1 반환
```

# 거듭제곱 계산
- a^n을 절반으로 쪼개서 계산, 시간복잡도 O(n log n)

```python
def power(a, n):
    result = 1             # 결과값 초기화
    while n > 0:           # 지수가 0보다 클 때까지 반복
        if n % 2 == 1:     # 지수가 홀수면 결과에 밑을 곱함
            result *= a
        a *= a             # 밑을 제곱하여 다음 단계로 이동
        n //= 2            # 지수를 반으로 나눔
    return result          # 계산된 결과 반환
```

# 병합 정렬
- 큰 배열을 반씩 계속 쪼개다가, 정렬하면서 다시 병합해서 전체를 정렬하는 방식의 알고리즘

```python
def merge_sort(arr):
    n = len(arr)           # 배열 길이
    size = 1               # 초기 병합 단위 크기 설정

    while size < n:        # 병합 단위가 배열 크기보다 작을 동안 반복
        for left in range(0, n, 2 * size):  # 병합할 구간의 시작 인덱스
            mid = min(left + size, n)       # 중간 인덱스
            right = min(left + 2 * size, n) # 오른쪽 끝 인덱스

            merged = []    # 병합 결과 담을 리스트
            i, j = left, mid  # 두 포인터 설정 (왼쪽, 오른쪽 구간 시작점)

            # 두 부분을 병합
            while i < mid and j < right:
                if arr[i] < arr[j]:
                    merged.append(arr[i])
                    i += 1
                else:
                    merged.append(arr[j])
                    j += 1

            merged += arr[i:mid]     # 왼쪽 나머지 추가
            merged += arr[j:right]   # 오른쪽 나머지 추가

            arr[left:left + len(merged)] = merged  # 병합 결과 원본에 반영

        size *= 2  # 병합 단위를 두 배로 증가
    return arr      # 정렬 완료된 배열 반환
```

# 최대값 찾기
- 배열을 반으로 쪼개서 최대값을 비교

```python
def find_max(arr, start, end):
    if start == end:              # 원소가 하나만 남은 경우
        return arr[start]         # 그 값을 반환 (기저 조건)

    mid = (start + end) // 2      # 중간 인덱스 계산
    left_max = find_max(arr, start, mid)     # 왼쪽 절반에서 최대값 재귀 호출
    right_max = find_max(arr, mid + 1, end)  # 오른쪽 절반에서 최대값 재귀 호출

    return max(left_max, right_max)  # 두 구간의 최대값 중 더 큰 값 반환
```

# 퀵 정렬 
- 기준점(pivot)을 기준으로 배열을 분할해서 정렬

```python
def quick_sort(arr):
    stack = [(0, len(arr) - 1)]  # 스택에 정렬할 구간 (start, end) 저장

    while stack:                 # 스택이 빌 때까지 반복
        start, end = stack.pop()  # 현재 구간 꺼냄
        if start >= end:
            continue             # 구간 크기가 1 이하이면 정렬할 필요 없음

        pivot = arr[start]       # 피벗은 구간의 첫 번째 원소로 설정
        low = start + 1
        high = end

        # 피벗 기준으로 양쪽에서 원소 이동
        while low <= high:
            while low <= high and arr[low] <= pivot:
                low += 1
            while low <= high and arr[high] > pivot:
                high -= 1
            if low < high:
                arr[low], arr[high] = arr[high], arr[low]  # 교환

        arr[start], arr[high] = arr[high], arr[start]  # 피벗을 제자리에 이동

        stack.append((start, high - 1))  # 왼쪽 구간 push
        stack.append((high + 1, end))    # 오른쪽 구간 push

    return arr  # 정렬 완료된 배열 반환
```

```python
print("이진 탐색:", binary_search([1, 3, 5, 7, 9], 5))  # ➜ 2
print("거듭제곱:", power(2, 10))                      # ➜ 1024
print("병합 정렬:", merge_sort([5, 1, 4, 2]))           # ➜ [1, 2, 4, 5]
print("퀵 정렬:", quick_sort([3, 6, 1, 8, 2]))           # ➜ [1, 2, 3, 6, 8]
```
